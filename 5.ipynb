```python
#5. Develop a knowledge base using propositional logic.

# Knowledge Base using Propositional Logic

# Facts stored in the knowledge base
facts = {
    "Raining": True,
    "Cloudy": True,
    "Windy": False
}

# Rules (Implications)
rules = [
    ("Raining", "Road_Wet"),
    ("Cloudy", "Rain_Possible"),
    ("Road_Wet", "Drive_Carefully")
]

# Function to perform inference
def infer(facts, rules):
    inferred = facts.copy()
    changed = True

    while changed:
        changed = False

        for premise, conclusion in rules:
            if inferred.get(premise, False) and not inferred.get(conclusion, False):
                inferred[conclusion] = True
                changed = True

    return inferred


# Perform inference
knowledge = infer(facts, rules)

# Display Knowledge Base
print("Knowledge Base")
print("-" * 30)

for fact, value in knowledge.items():
    print(f"{fact} : {value}")
```
