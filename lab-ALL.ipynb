```python
#2. Write a program to solve a problem using BFS and DFS.

from collections import deque

# Graph represented as an adjacency list
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['F'],
    'F': []
}

# Breadth-First Search (BFS)
def bfs(graph, start):
    visited = set()
    queue = deque([start])

    print("BFS Traversal:", end=" ")

    while queue:
        node = queue.popleft()

        if node not in visited:
            print(node, end=" ")
            visited.add(node)

            for neighbor in graph[node]:
                if neighbor not in visited:
                    queue.append(neighbor)

# Depth-First Search (DFS)
def dfs(graph, node, visited=None):
    if visited is None:
        visited = set()

    if node not in visited:
        print(node, end=" ")
        visited.add(node)

        for neighbor in graph[node]:
            dfs(graph, neighbor, visited)

# Main Program
print("Graph:", graph)

bfs(graph, 'A')

print("\nDFS Traversal:", end=" ")
dfs(graph, 'A')
```

```python
#5. Develop a knowledge base using propositional logic.

# Knowledge Base using Propositional Logic

# Facts stored in the knowledge base
facts = {
    "Raining": True,
    "Cloudy": True,
    "Windy": False
}

# Rules (Implications)
rules = [
    ("Raining", "Road_Wet"),
    ("Cloudy", "Rain_Possible"),
    ("Road_Wet", "Drive_Carefully")
]

# Function to perform inference
def infer(facts, rules):
    inferred = facts.copy()
    changed = True

    while changed:
        changed = False

        for premise, conclusion in rules:
            if inferred.get(premise, False) and not inferred.get(conclusion, False):
                inferred[conclusion] = True
                changed = True

    return inferred


# Perform inference
knowledge = infer(facts, rules)

# Display Knowledge Base
print("Knowledge Base")
print("-" * 30)

for fact, value in knowledge.items():
    print(f"{fact} : {value}")
```
